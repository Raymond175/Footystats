```python
import pandas as pd 
from footstatsapi.models import Fixture

data = pd.DataFrame(list(Fixture.objects.all().values("home_team_id", "away_team_id", "goals_home", "goals_away", "date", "league_id", "season")))

data.fillna(method="ffill", inplace=True)

data["result"] = (data["goals_home"] - data["goals_away"]).apply(lambda x: "H" if x > 0 else "A" if x < 0 else "D")
```

```python
def calculate_form(df, num_games=10):
    df = df.sort_values(by="date")
    df["points"] = df.apply(lambda row: 3 if row['home_score'] > row['away_score'] else (1 if row['home_score'] == row['away_score'] else 0), axis=1)
    
    df["home_form"] = df.groupby("home_team_id")["points"].transform(lambda x: x.rolling(window=num_games, min_periods=1).sum())
    df["away_form"] = df.groupby("away_team_id")["points"].transform(lambda x: x.rolling(window=num_games, min_periods=1).sum())
```
